```python
import pandas as pd
df=pd.read_csv("dataset.csv.csv")
df.head()
```

```python
df['Price range'].value_counts().sort_index()
```

```python
price_range_pct = df['Price range'].value_counts(normalize=True).sort_index() * 100
print(price_range_pct.round(2))
```

```python
import matplotlib.pyplot as plt

price_counts = df['Price range'].value_counts().sort_index()

plt.figure(figsize=(8,5))
price_counts.plot(kind='bar', color='mediumseagreen')
plt.title('Distribution of Price Ranges Among Restaurants')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Price range'], bins=[0.5,1.5,2.5,3.5,4.5], color='cornflowerblue', edgecolor='black')
plt.title('Histogram of Price Ranges')
plt.xlabel('Price Range')
plt.ylabel('Frequency')
plt.xticks([1,2,3,4])
plt.show()
```

```python
summary = pd.DataFrame({
    'Count': price_counts,
    'Percentage': price_range_pct.round(2)
})
print(summary)
```

 Nearly half of all restaurants (46.5%) fall into the lowest price range, and the distribution steadily drops as price increases — showing the dataset is dominated by budget-friendly restaurants, with premium (price range 4) restaurants being the rarest category.
