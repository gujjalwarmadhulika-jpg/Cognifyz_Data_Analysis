```python
df=pd.read_csv("dataset.csv.csv")
```

```python
print(df.head())
```

```python
print(df.columns)
```

```python
  top3 = df['Cuisines'].value_counts().head(3)

print(top3)
```

```python
percentage = (top3 / len(df)) * 100
print(percentage.round(2))
```

```python
 result = pd.DataFrame ( {
    "Cuisine": top3.index,
    "Restaurants": top3.values,
    "percentage": percentage.values.round(2)
} )

print(result)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(result["Cuisine"], result["percentage"], color=['#4C72B0', '#DD8452', '#55A868'])

plt.title("Top 3 Cuisines by Percentage of Restaurants")
plt.xlabel("Cuisine")
plt.ylabel("Percentage of Restaurants (%)")

# Add percentage labels on top of bars
for i, val in enumerate(result["percentage"]):
    plt.text(i, val + 0.2, f"{val}%", ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
